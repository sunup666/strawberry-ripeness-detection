```python
from ultralytics import YOLO

# 加载训练好的模型
model = YOLO('D:/FW/G17/LeTorch/runs/detect/train/weights/best.pt')

# 在验证集上评估并获取指标
metrics = model.val()  # 默认使用训练时的验证集

# 输出评估指标
print(f"Precision: {metrics.box.p}")    # 精准率
print(f"Recall: {metrics.box.r}")       # 召回率
print(f"F1-Score: {metrics.box.f1}")    # F1分数
```
